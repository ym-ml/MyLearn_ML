```python
import numpy as np
import pandas as pd
```

```python
data_xx=pd.DataFrame([[1,4],[2,1],[3,0],[4,1],[5,5]],columns=('x','y'))
data_xx
print(data_xx,"\n")

data_x_3=data_xx.copy()
data_x_3['x^2']=data_x_3['x']**2
data_x_3['x^3']=data_x_3['x']**3
data_x_3
print(data_x_3)
```

```python
data_xx['z']=1

data_x_3['z']=1
```

```python
data_xx=data_xx.reindex(['x','z','y'],axis=1)
print(data_xx)

data_x_3=data_x_3.reindex(['x','x^2','x^3','z','y'],axis=1)
print(data_x_3)
```

```python
X=data_xx.loc[:,'x':'z'].values
y=data_xx['y'].values

w=np.linalg.inv((X.T)@X)@(X.T)@y
j=np.array([[6,1],[-1,1]])#.reshape(2,3)
print(j.T,'\n\n',w.T)

w.T@j.T

```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data_xx.plot(kind='scatter',x='x',y='y')
x=np.arange(1,6)
y=w.T[0]*x+w.T[1]
plt.plot(x,y)
```

```python
X=data_x_3.loc[:,'x':'z'].values
y=data_x_3['y'].values

w=np.linalg.inv((X.T)@X)@(X.T)@y
j=np.array([[6,36,6**3,1],[-1,1,-1,1]])#.reshape(2,3)
print(j.T,'\n\n',w.T)

w.T@j.T

```
