```python
def get_result(w,test,test_columns):
    test_columns.append('prediction_result')
    
    ones=np.ones([test.shape[0],1])
    test=np.concatenate([ones,test],axis=1)
    #print(test,w)
    
    result=1/(1+np.exp(-test@w))
    result_class=(result>0.5).astype(int)    #not so good 1.=? 2.we need to select the max
    result_array=np.concatenate([test,result_class.T.reshape(len(result),1)],axis=1)
    result_frame=pd.DataFrame(result_array[:,1:],columns=test_columns)
    return (result,result_frame)
```
